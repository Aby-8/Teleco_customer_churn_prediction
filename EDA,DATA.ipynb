CHURN ANALYSIS

IMPORTING LIBRARIES


```python
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df.head()
```

```python
df.shape
df.dtypes
```

converting object into numeric(total charges should be numeric)

```python
dfe=df.copy()#copy of org dataset
df.TotalCharges=pd.to_numeric(df.TotalCharges,errors='coerce')
df.dtypes
df.isnull().sum()
```

checking for missing values

```python
df.loc[df['TotalCharges'].isnull()==True]
df.loc[df['TotalCharges'].isnull()==True]
```

handling missing values

```python
df.dropna(how='any',inplace=True)
df.loc[df['TotalCharges'].isnull()==True]
```

bins 

```python
df.tenure.max()
label=['{0}-{1}'.format(i,i+11) for i in range(1,72,12)]
df['tenure_grp']=pd.cut(df.tenure,range(1,80,12),right=False,labels=label)
df.tenure_grp.value_counts()
```

remove irrelevant columns

```python
df.drop(columns=['customerID','tenure'],axis=1,inplace=True)
df.head()
```

EDA

```python
for i,predictor in enumerate(df.drop(columns=['TotalCharges','MonthlyCharges','Churn'])):
    plt.figure(i)
    sns.countplot(data=df,x=predictor,hue='Churn')
```

encoding target into numeric and categorical into numeric

```python
df['Churn'] = np.where(df.Churn == 'Yes',1,0)
df.head()
```

```python
dfd=pd.get_dummies(df,dtype=int)
dfd.head()
```

```python
sns.kdeplot(data=dfd,x='MonthlyCharges',hue='Churn',fill=True,multiple='layer')
```

```python
plt.figure(figsize=(20,8))
dfd.corr()['Churn'].sort_values(ascending=False).plot(kind='bar')
```

```python
plt.figure()
sns.heatmap(dfd.corr(),cmap='Paired')
```
